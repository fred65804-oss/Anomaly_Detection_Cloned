```python
def loss(w):
    x = 2
    target = 10
    prediction = w * x
    error = (prediction - target)**2
    return error

weights = np.linspace(0,10,100) # 100 numbers between 0 and 10 with evenly spaced values
losses = [loss(w) for w in weights]

plt.figure(figsize = (10,6))
plt.plot(weights,losses)
plt.xlabel("Weights")
plt.ylabel("Losses for Weights")
plt.axvline(x = 5, color = 'r', linestyle = '--', label = 'Optimal w=5')
plt.grid(True)
plt.show()
```

```python
def gradient(w):
    """
        Loss = (Prediction - Target)**2
        dl/dW = 2*(Prediction - Target)*x
    """
    x = 2
    target = 10
    prediction = w*x
    loss = (prediction - target)**2
    return 2*(prediction - target)*x

test_points = [1,3,5,7,9]

g1,l1 = [],[]

for w in test_points:
    g = gradient(w)
    l = loss(w)
    g1.append(g)
    l1.append(l)

    print(f"Gradient: {g} Loss: {l}")
```

```python
print(g1)
print(l1)


plt.plot(g1,l1)
plt.xlabel("Gradients")
plt.ylabel("Loss")
plt.show()

```
